## 1. Contexto

A Rota Segura é um sistema de alerta de alagamento: sensores de chuva e de nível d'água indicam o risco de cada área, o sistema emite um alerta com a rota até o abrigo e a pessoa confirma com o check-in "Estou seguro".

Neste notebook, Data Science responde à pergunta **quando avisar**. Os modelos estimam o risco de alagamento a partir da chuva observada, e seus limiares viram a regra de risco baixo/médio/alto (RF04) usada pelo restante do sistema.

**Área de estudo:** A DEFINIR (bairro).

## 2. Perguntas de negócio

1. **Logística:** dada a chuva observada até o instante *t*, qual a probabilidade de haver alagamento na área nas próximas N horas?
2. **Linear:** dada a chuva observada até o instante *t*, qual valor a variável de intensidade (nível do córrego ou chuva acumulada) deve atingir nas próximas N horas?

**N (horas à frente):** A DEFINIR. Se a fonte de alagamentos trouxer só a data, o horizonte passa a ser o dia.

## 3. Momento da previsão

A previsão é feita no instante *t* e usa **somente o que o sistema já sabe em *t***: as variáveis olham para trás (por exemplo, chuva acumulada nas últimas horas) e o alvo olha para frente. Nenhuma informação posterior a *t* entra nas variáveis.

## 4. Variáveis-alvo

| Item | Regressão logística | Regressão linear |
|---|---|---|
| **Alvo** | `alagou`: houve alagamento registrado na área no horizonte | Nível do córrego, se houver estação hidrológica próxima; senão, chuva acumulada no horizonte |
| **Tipo** | Binário (0 ou 1) | Contínuo |
| **Unidade** | Sem unidade (0 ou 1) | cm (nível do córrego) ou mm (chuva acumulada) |
| **Horizonte** | Próximas N horas (ou o dia): A DEFINIR | Próximas N horas: A DEFINIR |
| **Fonte** | Ocorrências de alagamento: A DEFINIR (ver seção 5) | Estação hidrológica: A DEFINIR; ou INMET (chuva) |
| **Métricas** | Matriz de confusão, precisão, recall, F1 | MAE, RMSE, R² |

**Alvo da linear — regra de decisão:** se existir estação hidrológica perto do bairro, o alvo é o nível do córrego em cm. Sem ela, o alvo é a chuva acumulada nas próximas N horas, em mm, usando só o INMET.
